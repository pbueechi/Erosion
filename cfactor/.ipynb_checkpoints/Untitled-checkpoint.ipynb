```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
path = '/mnt/Data-Labo-RE/27_Natural_Resources-RE/321.4_WAUM_protected/Daten/Erosionsrisiko/C_Faktoren.csv'
df = pd.read_csv(path, encoding='latin-1', delimiter=';')

reg = "Tal"
method_a = f"{reg}_Pflug"
method_b = f"{reg}_Mulch"
method_c = f"{reg}_Direkt"

# Difference column (row-wise a - b); drop rows where either value is missing
diff_ab = (df[method_a] - df[method_b]).dropna()
diff_ac = (df[method_a] - df[method_c]).dropna()
diff_bc = (df[method_b] - df[method_c]).dropna()

data = [df[method_a].dropna(), df[method_b].dropna(), df[method_c].dropna(), diff_ab, diff_ac, diff_bc]
labels = [method_a, method_b, method_c, f"{method_a} - {method_b}", f"{method_a} - {method_c}", f"{method_b} - {method_c}"]

fig, ax = plt.subplots(figsize=(8, 5))
ax.boxplot(data, labels=labels, showmeans=True)  # use tick_labels= in Matplotlib >= 3.9
ax.axhline(0, color="grey", linestyle="--", linewidth=0.8)  # reference line for a - b
ax.set_ylabel("C Factor")
ax.grid(axis="y", alpha=0.3)
ax.tick_params(axis="x", labelrotation=45)

plt.tight_layout()
plt.show()
```

```python
mask = (df[method_a] - df[method_c]) < 0

# Values of c where a - b < 0
print(df.loc[mask, method_c])

# Full rows, with the difference added for context
result = df.loc[mask, ["Kultur AUI", method_a, method_c]].assign(diff=df[method_a] - df[method_c])
print(result.sort_values("diff"))

print(f"{mask.sum()} of {len(df)} rows have a - b < 0")
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D

def outlier_mask(s):
    """1.5 * IQR rule (same as a boxplot's whiskers)."""
    q1, q3 = s.quantile([0.25, 0.75])
    iqr = q3 - q1
    return (s < q1 - 1.5 * iqr) | (s > q3 + 1.5 * iqr)


df = df.dropna(subset=[method_a, method_c])
neg = (df[method_a] - df[method_c]) < 0
out = outlier_mask(df[method_a] - df[method_c])
normal = ~(neg | out)

fig, ax = plt.subplots(figsize=(9, 6))
ax.scatter(df.loc[normal, method_a], df.loc[normal, method_c], color="tab:blue",
           alpha=0.5, s=25, label="Other rows")
ax.scatter(df.loc[neg, method_a], df.loc[neg, method_c], color="tab:red",
           alpha=0.7, s=25, label=f"{method_a} - {method_c} < 0")
ax.scatter(df.loc[out, method_a], df.loc[out, method_c], facecolors="none",
           edgecolors="black", linewidths=1.5, s=90,
           label=f"Outlier {method_a} - {method_c} (1.5 × IQR)")

# label each marked point with its row index
for idx in df.index[neg | out]:
    ax.annotate(str(idx), (df.at[idx, method_a], df.at[idx, method_c]), xytext=(6, 0),
                textcoords="offset points", fontsize=7, va="center")

ax.set_title(f"{method_a} vs {method_c} (labels = row index)")
ax.set_xlabel(method_a)
ax.set_ylabel(method_c)
ax.tick_params(axis="y", labelrotation=45)
ax.grid(alpha=0.3)
lims = [min(ax.get_xlim()[0], ax.get_ylim()[0]),
        max(ax.get_xlim()[1], ax.get_ylim()[1])]
ax.plot(lims, lims, color="grey", linestyle="--", linewidth=1, label="x = y")
ax.set_xlim(lims)
ax.set_ylim(lims)

first_legend = ax.legend(loc="upper left")
ax.add_artist(first_legend)  # keep the first legend when adding a second one

labelled = df.index[neg | out]
handles = [Line2D([], [], linestyle="none") for _ in labelled]
texts = [f"{idx}: {df.at[idx, 'Kultur Kategorien 2020']}" for idx in labelled]

idx_legend = ax.legend(handles, texts, title="Kultur", loc="upper left",
                       bbox_to_anchor=(1.02, 1), fontsize=7,
                       handlelength=0, handletextpad=0)

for idx, text in zip(labelled, idx_legend.get_texts()):
    if neg[idx]:
        text.set_color("tab:red")

plt.tight_layout()
plt.show()

# Table of all labelled rows
# summary = df.assign(
#     diff=df[method_a] - df[method_b],
#     neg_diff=neg,
#     outlier_a=outlier_mask(df[method_a]),
#     outlier_c=outlier_mask(df[method_c]),
# )
# print(summary[neg | out])
```

```python
import seaborn as sns
import os
path = '~/mnt/eo-nas1/eoa-share/projects/028_Erosion/Erosion/cfactor'

comps = ['calibration_analysis_mlp_strat_noley/nn_predictions_per_pixel.csv', 'calibration_analysis_twobeta_noley/calibration_results_stratified_per_pixel.csv']

df_nn = pd.read_csv(os.path.join(path, comps[0]), encoding='latin-1', delimiter=',')
df_beta = pd.read_csv(os.path.join(path, comps[1]), encoding='latin-1', delimiter=',')
print(df_nn.shape, df_beta.shape)

df = df_nn.copy()
df.rename(columns={'C_pred': 'MLP_pred'}, inplace=True)
df['Beta_pred'] = df_beta['C_predicted']

# Convert from wide to long format
df_long = df.melt(
    id_vars='tillage',
    value_vars=['C_ref', 'MLP_pred', 'Beta_pred'],
    var_name='Type',
    value_name='C'
)

# Plot
plt.figure(figsize=(10, 6))
sns.boxplot(
    data=df_long,
    x='tillage',
    y='C',
    hue='Type',
)
plt.grid(True, axis="y", linestyle="--")

plt.xlabel('Tillage class')
plt.ylabel('C')
plt.title('Comparison of C_ref and C_pred by Tillage Class')
plt.legend(title='')
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

MIN_N = 10            # keep classes with > MIN_N paired observations
RESID_FACTOR = 1.5    # label classes with |mean residual| > RESID_FACTOR * RMSE

a = pd.read_csv('run_eb1/XY.csv')
df = pd.read_csv('run_eb1/XY_model_comparison_new.csv')
cols2add = a.columns[:6]
df[cols2add] = a[cols2add]
agg_col_name = "lnf_code"
setups = ["all", "all_till", "hq", "hq_till", "lq", "lq_till"]
cols = ["C_ref"] + setups

g = df.groupby(agg_col_name)
df_mean = g[cols].mean()
df_std = g[cols].std()

# per-subplot class selection: count rows where both C_ref and the setup are valid
keep = {}
for s in setups:
    n_pair = df.dropna(subset=["C_ref", s]).groupby(agg_col_name).size()
    keep[s] = n_pair[n_pair > MIN_N]

# shared axis limits from the plotted classes only (means ± std)
lo, hi = np.inf, -np.inf
for s, k in keep.items():
    if k.empty:
        continue
    for col in ["C_ref", s]:
        m = df_mean.loc[k.index, col]
        e = df_std.loc[k.index, col].fillna(0)
        lo = min(lo, np.nanmin(m - e))
        hi = max(hi, np.nanmax(m + e))
pad = 0.05 * (hi - lo)
lims = (lo - pad, hi + pad)

# palette over all codes so colours stay consistent across panels
codes = df_mean.index.tolist()
palette = dict(zip(codes, sns.color_palette("tab10", len(codes))))
plotted = sorted(set().union(*[k.index for k in keep.values()]))

fig, axes = plt.subplots(3, 2, figsize=(15, 14), sharex=True, sharey=True)

for ax, s in zip(axes.ravel(), setups):
    kc = keep[s].index
    ax.plot(lims, lims, "k--", lw=1)  # 1:1 line
    ax.set_xlim(lims)
    ax.set_ylim(lims)
    ax.set_aspect("equal")
    ax.set_title(s)
    ax.set_xlabel("C_ref (mean)")
    ax.set_ylabel(f"{s} (mean)")
    ax.grid(alpha=0.3)

    if len(kc) == 0:
        ax.text(0.5, 0.5, f"no classes with n > {MIN_N}",
                transform=ax.transAxes, ha="center")
        continue

    for c in kc:
        ax.errorbar(
            df_mean.loc[c, "C_ref"], df_mean.loc[c, s],
            xerr=df_std.loc[c, "C_ref"], yerr=df_std.loc[c, s],
            fmt="o", ms=7, color=palette[c], ecolor=palette[c],
            elinewidth=0.8, alpha=0.6, capsize=2,
        )

    # metrics on the kept class means
    d = df_mean.loc[kc, ["C_ref", s]].dropna()
    resid = d[s] - d["C_ref"]
    rmse = np.sqrt(np.mean(resid ** 2))
    r2 = np.corrcoef(d["C_ref"], d[s])[0, 1] ** 2 if len(d) > 1 else np.nan

    # label high residuals
    high = resid[resid.abs() > RESID_FACTOR * rmse]
    for c, r in high.items():
        x, y = d.loc[c, "C_ref"], d.loc[c, s]
        ax.scatter(x, y, s=130, facecolors="none", edgecolors="k",
                   linewidths=1.2, zorder=3)
        above = r > 0  # push the label away from the 1:1 line
        ax.annotate(
            f"{c} ({r:+.2f})", (x, y),
            xytext=(-8, 8) if above else (8, -8),
            textcoords="offset points",
            ha="right" if above else "left",
            va="bottom" if above else "top",
            fontsize=8, fontweight="bold",
            bbox=dict(facecolor="white", alpha=0.8, edgecolor="none", pad=1),
            arrowprops=dict(arrowstyle="-", lw=0.6, color="k"),
        )

    ax.text(
        0.04, 0.96,
        f"RMSE = {rmse:.2f}\nR² = {r2:.2f}\nn classes = {len(d)}\n"
        f"labelled: |res| > {RESID_FACTOR}·RMSE",
        transform=ax.transAxes, va="top", fontsize=9,
        bbox=dict(facecolor="white", alpha=0.7, edgecolor="none"),
    )

# shared legend: only classes that appear in at least one panel
n_total = g.size()
handles = [
    plt.Line2D([], [], marker="o", linestyle="", ms=7, color=palette[c],
               alpha=0.6, label=f"{c} (n = {n_total[c]})")
    for c in plotted
]
fig.legend(
    handles=handles, title=agg_col_name,
    loc="center left", bbox_to_anchor=(0.86, 0.5), frameon=False,
)

plt.tight_layout(rect=[0, 0, 0.85, 1])
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd

LONG = "data/all/s1_s2_cr_long_2021.parquet"
POSITIONS = [1, 100, 1000, 10000]     # n-th field in the file (0-based)
ORBIT = None                          # e.g. 88; None -> most frequent S1A orbit

long = pd.read_parquet(LONG)
long["date"] = pd.to_datetime(long["date"].astype(str), format="%Y%m%d")

s1 = long[long["platform"].astype(str)=="SENTINEL-1A"]
if ORBIT is None:
    ORBIT = int(s1["relative_orbit"].value_counts().idxmax())
s1 = s1[s1["relative_orbit"] == ORBIT]
direction = s1["direction"].iloc[0]
s2 = long[long["sensor"] == "S2"]

field_ids = sorted(long["field_id"].unique())
fields = [field_ids[p] for p in POSITIONS if p < len(field_ids)]
print(f"{len(field_ids)} fields; plotting {fields}; S1A {direction} orbit {ORBIT}")

twins = []
fig, axes = plt.subplots(len(fields), 1, figsize=(12, 3 * len(fields)), sharex=True)
for ax, fid in zip(axes, fields):
    a = s1[s1["field_id"] == fid].sort_values("date")
    b = s2[(s2["field_id"] == fid) & s2["B08"].notna()].sort_values("date")
    info = long.loc[long["field_id"] == fid, ["tillage", "betriebs_id"]].iloc[0]

    ax.plot(a["date"], a["cr_db"], "o-", ms=3, color="tab:blue", label=f"CR S1A {direction} {ORBIT}")
    ax.set_ylabel("CR (dB)", color="tab:blue")
    ax.tick_params(axis="y", colors="tab:blue")

    ax2 = ax.twinx()
    twins.append(ax2)
    ax2.plot(b["date"], b["B08"], "s-", ms=3, color="tab:green", label="S2 B08")
    ax2.set_ylabel("B08 (reflectance)", color="tab:green")
    ax2.tick_params(axis="y", colors="tab:green")

    ax.set_title(f"field {fid} | {info['tillage']} | farm {info['betriebs_id']} | "
                 f"S1: {a['cr_db'].notna().sum()} values, S2: {len(b)} clear dates", fontsize=10)
    ax.grid(alpha=0.3)

h1, l1 = axes[0].get_legend_handles_labels()
h2, l2 = twins[0].get_legend_handles_labels()
axes[0].legend(h1 + h2, l1 + l2, loc="upper left", fontsize=8, frameon=False)
plt.tight_layout()
plt.show()

fid = 669361
f = long[long["field_id"] == fid]
s1f = f[f["sensor"] == "S1"]

print(f"S1 rows: {len(s1f)}, S2 rows: {(f['sensor'] == 'S2').sum()}")
if len(s1f):
    print(s1f.groupby(["direction", "relative_orbit"])
          .agg(n=("cr_db", "size"), valid=("cr_db", lambda x: x.notna().sum()),
               n_inside=("n_inside", "first"),
               n_valid_median=("n_valid", "median"),
               n_used_median=("n_used", "median")))
```
